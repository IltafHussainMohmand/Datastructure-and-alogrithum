## Change List Item

```python
cars = ["Ford", "Volvo", "BMW"]
cars[1] = "toyota"
print(cars)

```

## Add an Item

```python
cars = ["Ford", "Volvo", "BMW"]
cars.append("toyota")
print(cars)
```

## Remove an Item

```python
cars = ["Ford", "Volvo", "BMW"]
cars.remove("Volvo")
print(cars)
```
